```python
class MyCalendar:
    def __init__(self):
        self.range = [[0, 10**9]]


    def book(self, startTime, endTime):
        # Check if everything is booked
        if not self.range:
            return False

        index = -1
        left = 0
        right = len(self.range) - 1

        while left <= right:
            mid = (left + right) // 2

            if (
                self.range[mid][0] <= startTime <= self.range[mid][1] or
                self.range[mid][0] <= endTime - 1 <= self.range[mid][1]
            ):
                index = mid
                break

            elif  self.range[mid][1] < startTime:
                left = mid + 1

            else:
                right = mid - 1

        # Check if it's bookable.
        if (
            index == -1 or
            not (self.range[index][0] <= startTime <= self.range[index][1] and
                 self.range[index][0] <= endTime - 1 <= self.range[index][1])
        ):
            return False


        prevStart, prevEnd = self.range[index][0], self.range[index][1]

        # Remove this interval from the list
        self.range.pop(index)

        if not (startTime == prevStart and endTime - 1 == prevEnd):
            if startTime == prevStart:
                self.range.insert(index, [endTime, prevEnd])

            elif endTime - 1 == prevEnd:
                self.range.insert(index, [prevStart, startTime - 1])

            else:
                self.range.insert(index, [prevStart, startTime - 1])
                self.range.insert(index + 1, [endTime, prevEnd])

        return True
```
