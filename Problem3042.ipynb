```python
class Solution:
    def countPrefixSuffixPairs(self, words):
        # Naive Approach
        # This works because len(words) <= 50
        result = 0
        n = len(words)
        for i in range(n):
            for j in range(i + 1, n):
                n1 = len(words[i])
                n2 = len(words[j])
                is_valid = True if n2 >= n1 else False

                # Check if it's prefix
                if is_valid:
                    for k in range(n1):
                        if words[i][k] != words[j][k]:
                            is_valid = False
                            break

                # Check if it's suffix
                if is_valid:
                    for k in range(n1):
                        if words[i][n1 - 1 - k] != words[j][n2 - 1 - k]:
                            is_valid = False
                            break

                if is_valid:
                    result += 1

        return result
```

```python
solution = Solution()
```

```python
words = ["a","aba","ababa","aa"]

print(solution.countPrefixSuffixPairs(words))
```
