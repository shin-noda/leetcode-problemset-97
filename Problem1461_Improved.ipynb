```python
class Solution:
    def hasAllCodes(self, s, k):
        # Idea: Use Rolling Hash

        # 2^k distinct codes
        total_needed = 1 << k

        # Early exit guard
        if len(s) < total_needed + k - 1:
            return False

        # Compute the bitmask of k ones: (2^k) - 1
        # For k = 3: (1 << 3) - 1 = 8 - 1 = 7 = 0b111
        bitmask = total_needed - 1

        # Calculate the rolling hash for the very first window s[0 : k]
        rolling_hash = 0
        for i in range(k):
            rolling_hash = (rolling_hash << 1) | int(s[i])

        seen = {rolling_hash}

        # Slide the window from index k to len(s) - 1 in O(1) per step
        for i in range(k , len(s)):
            # Shift left, add incoming bit, and mask out the overflow bit
            rolling_hash = ((rolling_hash << 1) & bitmask) | int(s[i])
            seen.add(rolling_hash)

            # Early exit if full coverage achieved
            if len(seen) == total_needed:
                return True

        return False
```

```python
solution = Solution()
```

```python
s = "00110110"
k = 2

print(solution.hasAllCodes(s, k))
```
