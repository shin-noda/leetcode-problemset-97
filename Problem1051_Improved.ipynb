```python
class Solution:
    def heightChecker(self, heights):
        RANGE = 101
        n = len(heights)

        # From height 1 to 100, 0-indexed
        counts = [0] * RANGE

        for h in heights:
            counts[h] += 1

        # counts[0] is always 0, so we start with counts[1] which may not be 0.
        index = 1
        checked = 0
        result = 0
        while checked < n:
            while index < RANGE and counts[index] <= 0:
                index += 1

            if heights[checked] != index:
                result += 1

            checked += 1
            counts[index] -= 1

        return result
```
