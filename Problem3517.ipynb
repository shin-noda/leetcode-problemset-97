```python
class Solution:
    def smallestPalindrome(self, s):
        # Idea: if counts[i] is an odd number, but a single character in the middle.
        # For other charaters, process from a to z.

        # 26 alphabets
        counts = [0] * 26
        for alphabet in s:
            counts[ord(alphabet) - ord('a')] += 1

        half = []
        middle = []
        for i in range(len(counts)):
            if counts[i] == 0:
                continue

            # We found the middle character
            if counts[i] % 2 != 0:
                middle.append(chr(ord('a') + i))
                counts[i] -= 1

            half.append(chr(ord('a') + i) * (counts[i] // 2))

        return "".join(half) + "".join(middle) + "".join(half[::-1])
```
