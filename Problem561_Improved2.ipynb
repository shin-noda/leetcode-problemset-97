```python
class Solution:
    def arrayPairSum(self, nums):
        # Use Counting Sort
        # Time Complexity: O(K) where n = len(nums) and K = 20001
        OFFSET = 10000
        k = 20001

        count = [0] * k
        for num in nums:
            count[num + OFFSET] += 1

        total_sum = 0
        take = True

        for i in range(k):
            if count[i] == 0:
                continue

            val = i - OFFSET

            # If 'take' is True, we take ceil(count / 2) items.
            # If 'take' is False, we take floor(count / 2) items.
            num_to_take = (count[i] + (1 if take else 0)) // 2
            total_sum += num_to_take * val

            # If the frequency is odd, the parity state toggles for the next value
            if count[i] % 2 != 0:
                take = not take

        return total_sum
```
