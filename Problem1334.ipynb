```python
import heapq

class Solution:
    def findTheCity(self, n, edges, distanceThreshold):
        graph = {}

        # Initialize graph
        for i in range(n):
            for j in range(n):
                if j == i:
                    continue

                if i not in graph:
                    graph[i] = {}

                if j not in graph:
                    graph[j] = {}

                graph[i][j] = float('inf')
                graph[j][i] = float('inf')

        remaining = set()
        for i in range(n):
            remaining.add(i)

        for edge in edges:
            start, end, weight = edge

            graph[start][end] = weight
            graph[end][start] = weight

            # Remove 'start' and 'end' from 'remaining.
            # If it doesn't exist, ignore it.
            remaining.discard(start)
            remaining.discard(end)

        # If there is any element remaind in 'remaining',
        # we can simply return the biggest remaining city.
        if remaining:
            return max(list(remaining))

        # Run Dijkstra for all nodes
        # O(V * (VlogE))
        for start in graph.keys():
            min_heap = [(0, start)]
            visited = set()

            while min_heap:
                w, curr = heapq.heappop(min_heap)

                if curr in visited:
                    continue

                # Unreacheable
                if w == float('inf'):
                    break

                # Update the weight
                if curr != start:
                    graph[start][curr] = w

                # Mark as visited
                visited.add(curr)

                if len(visited) == n:
                    break

                # Explore neighbours
                for neighbour in graph[curr].keys():
                    weight = graph[curr][neighbour]

                    if neighbour not in visited:
                        heapq.heappush(min_heap, (w + weight, neighbour))

        min_count = float('inf')
        max_city = -1
        for start in range(n):
            cities = []

            for dest in range(n):
                if dest == start:
                    continue

                if graph[start][dest] <= distanceThreshold:
                    cities.append(dest)

            if len(cities) <= min_count:
                max_city = start
                min_count = len(cities)

        return max_city
```

```python
solution = Solution()
```

```python
n = 4
edges = [[0,1,3],[1,2,1],[1,3,4],[2,3,1]]
distanceThreshold = 4

print(solution.findTheCity(n, edges, distanceThreshold))
```
