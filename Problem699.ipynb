```python
class Node:
    def __init__(self):
        self.left = None
        self.right = None
        self.max_val = 0
        self.lazy = 0


class SegmentTree:
    def __init__(self):
        self.root = Node()


    def query(self, node, l, r, ql, qr):
        # Completely outside
        if qr <= l or r <= ql:
            return 0

        # Completely inside
        if ql <= l and r <= qr:
            return node.max_val

        mid = (l + r) // 2

        # Push lazy value to children
        self.push(node)

        left_max = self.query(node.left, l, mid, ql, qr)
        right_max = self.query(node.right, mid, r, ql, qr)

        return max(left_max, right_max)


    def update(self, node, l, r, ql, qr, height):
        # Completely outside
        if qr <= l or r <= ql:
            return

        # Completely inside
        if ql <= l and r <= qr:
            node.max_val = height
            node.lazy = height
            return

        mid = (l + r) // 2

        self.push(node)

        self.update(node.left, l, mid, ql, qr, height)
        self.update(node.right, mid, r, ql, qr, height)

        node.max_val = max(node.left.max_val, node.right.max_val)


    def push(self, node):
        if node.left is None:
            node.left = Node()

        if node.right is None:
            node.right = Node()

        if node.lazy:
            node.left.max_val = node.lazy
            node.left.lazy = node.lazy

            node.right.max_val = node.lazy
            node.right.lazy = node.lazy

            node.lazy = None


class Solution:
    def fallingSquares(self, positions):
        # answer[i] represents the height described above after dropping the ith square.
        answer = []
        max_height = 0

        st = SegmentTree()

        for left, side_length in positions:
            right = left + side_length

            # Find the highest point underneath the new square.
            base_height = st.query(st.root, 0, 10**9, left, right)

            # The square lands on that height.
            new_height = base_height + side_length

            # The entire square now has this top height.
            st.update(st.root, 0, 10**9, left, right, new_height)

            max_height = max(max_height, new_height)
            answer.append(max_height)

        return answer
```

```python
solution = Solution()
```

```python
positions = [[1,2],[2,3],[6,1]]

print(solution.fallingSquares(positions))
```

```python
positions = [[100,100],[200,100]]

print(solution.fallingSquares(positions))
```
