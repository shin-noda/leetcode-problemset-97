```python
import heapq

class Solution:
    def maxProbability(self, n, edges, succProb, start_node, end_node):
        # Idea: Use Max-Heap because we want to maximize the probility, not minimize.
        # Note: n is not always equal to len(edges) in this problem.

        # Create a graph
        adj = [[] for _ in range(n)]
        for i in range(len(edges)):
            u, v, p = edges[i][0], edges[i][1], succProb[i]
            adj[u].append((-p, v))
            adj[v].append((-p, u))

        # (probability, current_node)
        # The probability of moving from node i to node i is 1.0 (it's -1.0 because it's max_heap).
        max_heap = [(-1.0, start_node)]
        visited = set()

        while max_heap:
            prob, curr = heapq.heappop(max_heap)

            # Make it positive
            prob *= -1.0

            # If it's visited, we skip it.
            if curr in visited:
                continue

            # Add to the visited set.
            visited.add(curr)

            # If we've reached the goal, return the probability.
            if curr == end_node:
                return prob

            # Explore its neighbours
            for probability, neighbour in adj[curr]:
                if neighbour not in visited:
                    heapq.heappush(max_heap, (probability * prob, neighbour))

        # Unreachable
        return 0.0
```

```python
solution = Solution()
```

```python
n = 3
edges = [[0,1],[1,2],[0,2]]
succProb = [0.5,0.5,0.2]
start_node = 0
end_node = 2

print(solution.maxProbability(n, edges, succProb, start_node, end_node))
```
