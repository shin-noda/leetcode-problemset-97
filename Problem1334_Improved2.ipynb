```python
class Solution:
    def findTheCity(self, n, edges, distanceThreshold):
        # Use Floyd-Warshall.
        # Time Complexity: O(n^3)

        # Initialize dist matrix with infinity
        dist = [[float('inf')] * n for _ in range(n)]

        for i in range(n):
            # The shortest from city i to city i is 0, not float('inf').
            dist[i][i] = 0

        for u, v, w in edges:
            dist[u][v] = w
            dist[v][u] = w

        # Floyd-Warshall: k is the intermediate node
        for k in range(n):
            for i in range(n):
                for j in range(n):
                    if dist[i][k] + dist[k][j] < dist[i][j]:
                        dist[i][j] = dist[i][k] + dist[k][j]

        # Count cities within threshold
        min_reachable = float('inf')
        result_city = -1

        for i in range(n):
            reachable = 0
            for j in range(n):
                # We don't want to count the starting city itself
                if i == j:
                    continue

                # If the city is within the threshold, add it to our count
                if dist[i][j] <= distanceThreshold:
                    reachable += 1

            # "<=" automatically picks the larger index on tie
            if reachable <= min_reachable:
                min_reachable = reachable
                result_city = i

        return result_city
```

```python
solution = Solution()
```

```python
n = 4
edges = [[0,1,3],[1,2,1],[1,3,4],[2,3,1]]
distanceThreshold = 4

print(solution.findTheCity(n, edges, distanceThreshold))
```
