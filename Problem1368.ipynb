```python
from collections import deque

class Solution:
    def minCost(self, grid):
        # 0-1 BFS with deque
        # Time Complexity: O(m x n)
        rows, cols = len(grid), len(grid[0])

        # 1: Right, 2: Left, 3: Down, 4: Up
        directions = [(0, 1), (0, -1), (1, 0), (-1, 0)]

        # Distance array to keep track of minimum cost to reach each cell
        dist = [[float('inf')] * cols for _ in range(rows)]
        dist[0][0] = 0

        # Double-ended queue for 0-1 BFS: stores (r, c)
        queue = deque([(0, 0)])

        while queue:
            r, c = queue.popleft()

            # If we reached the bottom-right cell, return its minimal cost
            if r == rows - 1 and c == cols - 1:
                return dist[r][c]

            # Try moving in all 4 possible directions
            for idx, (dr, dc) in enumerate(directions, start=1):
                nr, nc = r + dr, c + dc

                if 0 <= nr < rows and 0 <= nc < cols:
                    # Cost is 0 if moving in arrow's direction, else 1
                    weight = 0 if grid[r][c] == idx else 1
                    new_cost = dist[r][c] + weight

                    if new_cost < dist[nr][nc]:
                        dist[nr][nc] = new_cost

                        if weight == 0:
                            # Priority execution for free moves
                            queue.appendleft((nr, nc))

                        else:
                            queue.append((nr, nc))

        dist[rows - 1][cols - 1]
```

```python
solution = Solution()
```
