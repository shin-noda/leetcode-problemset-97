```python
class Solution:
    def minSwapsCouples(self, row):
        answer = 0

        # Pairs are always 2n and 2n + 1, not 2n and 2n -1.
        # e.g. 2 and 3 -> valid pair. 2 and 1 -> not a valid pair.
        num_len = len(row) // 2
        pairs = []

        # First, eliminate valid pairs
        for i in range(num_len):
            index = 2 * i

            # We want 'a' to be even
            a, b = row[index], row[index + 1]
            if b % 2 == 0:
                a, b = b, a

            if b != a + 1:
                pairs.append([a, b])

        # Next, swap pairs.
        while pairs:
            # We always pop the last element of pairs and if 'is_doublepop' is true,
            # we pop the list twice.
            is_found = False
            is_doublepop = False
            indices = []
            n = len(pairs)

            for i in range(1, n):
                first, second = pairs[i][0], pairs[i][1]

                if pairs[0][0] % 2 == 0:
                    if pairs[0][0] + 1 == first:
                        indices.append(i)
                        indices.append(0)
                        is_found = True

                    elif pairs[0][0] + 1 == second:
                        indices.append(i)
                        indices.append(1)
                        is_found = True

                else:
                    if pairs[0][0] - 1 == first:
                        indices.append(i)
                        indices.append(0)
                        is_found = True

                    elif pairs[0][0] - 1 == second:
                        indices.append(i)
                        indices.append(1)
                        is_found = True

                # Check if a double-swap is possible
                if is_found:
                    if pairs[0][1] % 2 == 0:
                        if pairs[0][1] + 1 == first:
                            indices.append(i)
                            indices.append(0)
                            is_doublepop = True

                        elif pairs[0][1] + 1 == second:
                            indices.append(i)
                            indices.append(1)
                            is_doublepop = True

                    else:
                        if pairs[0][1] - 1 == first:
                            indices.append(i)
                            indices.append(0)
                            is_doublepop = True

                        elif pairs[0][1] - 1 == second:
                            indices.append(i)
                            indices.append(1)
                            is_doublepop = True

                    break

            if is_doublepop:
                # Check if the popped element is already in the last element
                if indices[0] == n - 1:
                    # Swap the first element with the second last element
                    pairs[0], pairs[-2] = pairs[-2], pairs[0]

                else:
                    # Swap one element with the last element
                    pairs[0], pairs[-1] = pairs[-1], pairs[0]

                    # Swap another element with the second last element
                    # Note: we uses indices[2], not indices[1]
                    pairs[indices[2]], pairs[-2] = pairs[-2], pairs[indices[2]]

                # Double pop
                pairs.pop()
                pairs.pop()

            else:
                # We keep the first element of the first index,
                # so we swap the second element of the first index.
                pairs[0][1], pairs[indices[0]][indices[1]] = pairs[indices[0]][indices[1]], pairs[0][1]

                # Swap the first and the last elements
                pairs[0], pairs[-1] = pairs[-1], pairs[0]

                # Single pop
                pairs.pop()

            answer += 1

        return answer
```
