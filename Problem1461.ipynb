```python
from collections import deque

class Solution:
    def hasAllCodes(self, s, k):
        # Idea: Sliding Window of the size k with deque.
        combination = set()
        dq = deque()
        n = len(s)

        # Quick check
        # De Bruijn sequence
        if n < 2**k + k - 1:
            return False

        # Initialize dq
        for i in range(k):
            dq.append(s[i])

        # Add the first combination
        combination.add("".join(list(dq)))

        for i in range(k, n):
            # Remove the left edge
            dq.popleft()

            # Add a new value to the right edge
            dq.append(s[i])

            # O(20) = O(1) since k <= 20
            combination.add("".join(list(dq)))

        return len(combination) == 2**k
```

```python
solution = Solution()
```

```python
s = "00110110"
k = 2

print(solution.hasAllCodes(s, k))
```
