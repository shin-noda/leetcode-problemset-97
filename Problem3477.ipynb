```python
class Solution:
    def numOfUnplacedFruits(self, fruits, baskets):
        answer = 0
        n = len(fruits)

        for i in range(n):
            for j in range(n):
                if baskets[j] >= fruits[i]:
                    fruits[i] = 0
                    baskets[j] = 0
                    break

            # Check if we could put fruits[i] into a basket
            if fruits[i] != 0:
                answer += 1

        return answer
```

```python
solution = Solution()
```

```python
fruits = [4,2,5]
baskets = [3,5,4]

print(solution.numOfUnplacedFruits(fruits, baskets))
```
