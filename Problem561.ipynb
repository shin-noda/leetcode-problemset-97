```python
class Solution:
    def arrayPairSum(self, nums):
        # Naive approach.
        # Sort the array in increasing order and group by 2.
        nums.sort()

        sum_val = 0
        for i in range(0, len(nums), 2):
            sum_val += min(nums[i], nums[i + 1])

        return sum_val
```
