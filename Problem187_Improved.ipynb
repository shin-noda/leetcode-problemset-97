```python
class Solution:
    def findRepeatedDnaSequences(self, s):
        # Quick Check
        if len(s) <= 10:
            return []

        # Use Rolling Hash
        MOD = 10**9 + 7
        BASE = 31

        n = len(s)
        n_dna = 10
        hash = 0

        # Note: it's a power of (n_dna - 1), NOT n_dna
        power = pow(BASE, n_dna - 1, MOD)
        sequences = []

        # hash-> (count, index)
        # Note: we store index for the collision verification
        dna_map = {}

        # Initialize the hash
        for i in range(n_dna):
            hash = (hash * BASE + (ord(s[i]) - ord('A') + 1)) % MOD

        # First 10-letter-long DNA sequence with a count of 1 with index 0
        dna_map[hash] = (1, 0)

        for i in range(1, n - n_dna + 1):
            count = 0

            # Subtract the hash value from the leftmost DNA character
            hash = (hash - (power * (ord(s[i - 1]) - ord('A') + 1))) % MOD

            # Add the hash value of the rightmost DNA sequence
            hash = (hash * BASE + (ord(s[i + n_dna - 1])  - ord('A') + 1)) % MOD

            # Check if the count is exactly 1 and the characters match
            if hash in dna_map:
                count, index = dna_map[hash]

                # Collision checking
                if count == 1 and s[index:index + n_dna] == s[i:i + n_dna]:
                    sequences.append(s[index:index + n_dna])

            # Increment the count here, after checking
            count += 1
            dna_map[hash] = (count, i)

        return sequences
```

```python
solution = Solution()
```
