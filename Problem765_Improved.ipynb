```python
class Solution:
    def minSwapsCouples(self, row):
        answer = 0

        for i in range(0, len(row), 2):
            partner = row[i] ^ 1

            j = row.index(partner, i + 1)

            if j != i + 1:
                row[i + 1], row[j] = row[j], row[i + 1]
                answer += 1

        return answer
```

```python
solution = Solution()
```

```python
row = [0,2,1,3]

print(solution.minSwapsCouples(row))
```
