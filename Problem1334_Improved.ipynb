```python
from collections import defaultdict
import heapq

class Solution:
    def findTheCity(self, n, edges, distanceThreshold):
        # Time Complexity: O(V ElogV) where V = number of cities n and E is the number of edges
        adj = defaultdict(list)
        for u, v, w in edges:
            adj[u].append((v, w))
            adj[v].append((u, w))


        def get_reachable_count(start):
            dist = [float('inf')] * n
            dist[start] = 0
            min_heap = [(0, start)]

            while min_heap:
                d, u = heapq.heappop(min_heap)

                if d > dist[u]:
                    continue

                # Early stop: don't expand neighbours if distance already exceeds threshold.
                if d > distanceThreshold:
                    break

                for v, weight in adj[u]:
                    if dist[u] + weight < dist[v]:
                        dist[v] = dist[u] + weight
                        heapq.heappush(min_heap, (dist[v], v))

            # Count reachable cities (excluding self)
            reachable_count = 0
            for i in range(n):
                # We don't want to count the starting city itself
                if i == start:
                    continue

                # If the city is within the threshold, add it to our count
                if dist[i] <= distanceThreshold:
                    reachable_count += 1

            return reachable_count

        min_count = float('inf')
        result = -1

        for i in range(n):
            count = get_reachable_count(i)

            if count <= min_count:
                min_count = count
                result = i

        return result
```

```python
solution = Solution()
```

```python
n = 4
edges = [[0,1,3],[1,2,1],[1,3,4],[2,3,1]]
distanceThreshold = 4

print(solution.findTheCity(n, edges, distanceThreshold))
```
