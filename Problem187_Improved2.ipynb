```python
class Solution:
    def findRepeatedDnaSequences(self, s):
        # Quick Check
        if len(s) <= 10:
            return []

        mapping = {'A': 0, 'C': 1, 'G': 2, 'T': 3}

        # Keep only the last 20 bits
        # For A, C, G, and T, it can be represented as 00, 01, 10, and 11.
        # So, 2 (bits) x 10 (DNA characters) = 20 (bits * DNA characters)
        #
        # (1 << 20) -> Move the binary 1 left by 20 positions
        # -> 100000000000000000000 (1 followed by 20 zeros, 21 digits)
        # Then, subtract 1 in binary -> 011111111111111111111
        # So 'bitmask' is 11111111111111111111 (20 digits, 20 ones)
        bitmask = (1 << 20) - 1

        bit_hash = 0
        seen = set()
        repeated = set()

        # Build initial 10-character hash
        for i in range(10):
            bit_hash = (bit_hash << 2) | mapping[s[i]]

        seen.add(bit_hash)

        # Roll the 20-bit window across the string
        for i in range(10, len(s)):
            # Left shift 2 bits, add new char, mask out bits beyond 20
            bit_hash = ((bit_hash << 2) | mapping[s[i]]) & bitmask

            if bit_hash in seen:
                repeated.add(s[i - 9 : i + 1])

            else:
                seen.add(bit_hash)

        return list(repeated)
```

```python
solution = Solution()
```

```python
s = "AAAAACCCCCAAAAACCCCCCAAAAAGGGTTT"

print(solution.findRepeatedDnaSequences(s))
```
