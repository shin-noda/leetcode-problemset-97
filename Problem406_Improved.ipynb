```python
class FenwickTree:
    def __init__(self, size):
        self.size = size
        self.tree = [0] * (size + 1)

        # Initialize all positions as 1 (representing an empty slot)
        for i in range(1, size + 1):
            self.update(i, 1)


    def update(self, i, delta):
        while i <= self.size:
            self.tree[i] += delta
            i += i & (-i)


    def find_kth_empty(self, k):
        # Binary lifting on Fenwick Tree to find the index of the k-th empty slot i in O(log N)
        i = 0
        mask = 1 << (self.size.bit_length() - 1)

        while mask > 0:
            next_i = i + mask

            if next_i <= self.size and self.tree[next_i] < k:
                i = next_i
                k -= self.tree[i]

            mask >>= 1

        # 1-based index to 1-based position
        return i + 1


class Solution:
    def reconstructQueue(self, people):
        n = len(people)

        # Sort by height ascending, then k descending
        people.sort(key=lambda x: (x[0], -x[1]))

        result = [None] * n
        bit = FenwickTree(n)

        for person in people:
            k = person[1]

            # We need the (k + 1)-th empty slot
            position = bit.find_kth_empty(k + 1)

            # Place person at position - 1 (0-indexed)
            result[position - 1] = person

            # Mark position as filled by reducing its count
            bit.update(position, -1)

        return result
```

```python
solution = Solution()
```

```python
people = [[7,0],[4,4],[7,1],[5,0],[6,1],[5,2]]

print(solution.reconstructQueue(people))
```
