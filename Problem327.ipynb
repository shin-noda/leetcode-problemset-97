```python
class Solution:
    def merge(self, left, right):
        i = j = 0
        merged = []

        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                merged.append(left[i])
                i += 1

            else:
                merged.append(right[j])
                j += 1

        # Exhaust the rest of elements explicitly
        merged.extend(left[i:])
        merged.extend(right[j:])

        return merged


    def countRange(self, left, right):
        # Note: Prefix Sum Arrays of 'left' and 'right' are both sorted
        # -> Use this monotonicity and Two Pointer solution.
        count = start = end = 0
        for l in left:
            while start < len(right) and right[start] - l < self.lower:
                start += 1

            while end < len(right) and right[end] - l <= self.upper:
                end += 1

            count += end - start

        return count


    def mergeSort(self, arr):
        if len(arr) <= 1:
            return arr, 0

        mid = len(arr) // 2

        left, left_count = self.mergeSort(arr[:mid])
        right, right_count = self.mergeSort(arr[mid:])

        # Count valid range sum
        range_sum_count = self.countRange(left, right)

        merged = self.merge(left, right)

        return merged, (left_count + right_count + range_sum_count)


    def countRangeSum(self, nums, lower, upper):
        # Use Merge Sort with Prefix Sum
        self.lower = lower
        self.upper = upper

        P = [0]
        for num in nums:
            P.append(P[-1] + num)

        _, count = self.mergeSort(P)

        return count
```

```python
solution = Solution()
```

```python
nums = [-2,5,-1]
lower = -2
upper = 2

print(solution.countRangeSum(nums, lower, upper))
```

```python
nums = [0]
lower = 0
upper = 0

print(solution.countRangeSum(nums, lower, upper))
```
