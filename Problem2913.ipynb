```python
class Solution:
    def sumCounts(self, nums):
        n = len(nums)
        count = 0

        for i in range(n):
            for j in range(n - i):
                count += len(set(nums[j : j + i + 1]))**2

        return count
```

```python
solution = Solution()
```

```python
nums = [1,2,1]

print(solution.sumCounts(nums))
```

```python
nums = [1,1]

print(solution.sumCounts(nums))
```
