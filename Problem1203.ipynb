```python
from collections import deque, defaultdict

class Solution:
    def sortItems(self, n, m, group, beforeItems):
        # Assign unique group IDs to all items with group == -1
        group_id = m
        for i in range(n):
            if group[i] == -1:
                group[i] = group_id
                group_id += 1

        # Build graph structures for both items and groups
        item_graph = defaultdict(list)
        item_indegree = [0] * n

        group_graph = defaultdict(list)
        group_indegree = [0] * group_id

        for v in range(n):
            for u in beforeItems[v]:
                # Edge u -> v in item graph
                item_graph[u].append(v)
                item_indegree[v] += 1

                # If they belong to different groups, add edge in group graph
                if group[u] != group[v]:
                    group_graph[group[u]].append(group[v])
                    group_indegree[group[v]] += 1


        # Helper function for standard Kahn's Algorithm
        def topo_sort(nodes, graph, indegree):
            queue = deque([node for node in nodes if indegree[node] == 0])
            order = []

            while queue:
                curr = queue.popleft()
                order.append(curr)

                for neighbour in graph[curr]:
                    indegree[neighbour] -= 1

                    if indegree[neighbour] == 0:
                        queue.append(neighbour)

            return order if len(order) == len(nodes) else []


        # Perform topological sort on both levels
        all_groups = list(range(group_id))
        all_items = list(range(n))

        sorted_groups = topo_sort(all_groups, group_graph, group_indegree)
        sorted_items = topo_sort(all_items, item_graph, item_indegree)

        # If either graph has a cycle, ordering is impossible
        if not sorted_groups or not sorted_items:
            return []

        # Group sorted items by their group ID
        items_by_group = defaultdict(list)
        for item in sorted_items:
            items_by_group[group[item]].append(item)

        # Reconstruct final array following group order
        result = []
        for g in sorted_groups:
            result.extend(items_by_group[g])

        return result
```

```python
solution = Solution()
```

```python
n = 8
m = 2
group = [-1,-1,1,0,0,1,0,-1]
beforeItems = [[],[6],[5],[6],[3,6],[],[],[]]

print(solution.sortItems(n, m, group, beforeItems))
```
