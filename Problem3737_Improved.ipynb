```python
class Solution:
    def countMajoritySubarrays(self, nums, target):
        # Use Merge Sort (O(NlogN))

        # Step 1: Build prefix sum array with +1 / -1 values
        # Note: P[0] = 0 always. So if nums is not empty, len(P) is at least 2.
        P = [0]
        curr = 0
        for x in nums:
            curr += 1 if x == target else -1
            P.append(curr)

        # Step 2: Merge Sort to count pairs (i, j) with i < j and P[i] < P[j]
        def mergeSort(arr):
            if len(arr) <= 1:
                return arr, 0

            mid = len(arr) // 2
            left, count_left = mergeSort(arr[:mid])
            right, count_right = mergeSort(arr[mid:])

            merged = []
            i = j = 0
            count = count_left + count_right

            while i < len(left) and j < len(right):
                if left[i] < right[j]:
                    # left[i] is smaller than right[j] and every element after right[j]
                    count += len(right) - j
                    merged.append(left[i])
                    i += 1

                else:
                    merged.append(right[j])
                    j += 1

            merged.extend(left[i:])
            merged.extend(right[j:])

            return merged, count

        _, total_subarray = mergeSort(P)

        return total_subarray
```
