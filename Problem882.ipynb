```python
import heapq

class Solution:
    def reachableNodes(self, edges, maxMoves, n):
        # Idea: Split the logic into two clear phases:
        # Phase 1: Standard Dijkstra
        # Phase 2: Count Reachable Nodes

        # Time Complexity: O(ElogV) where V = n and E = len(edges)
        # Space Complexity: O(V + E)

        # Build adjacency list: (neighbour, weight)
        adj = {i: [] for i in range(n)}
        for u, v, cnt in edges:
            adj[u].append((v, cnt + 1))
            adj[v].append((u, cnt + 1))

        # Phase 1: Dijkstra to find shortest distances to original nodes
        dist = {i: float('inf') for i in range(n)}
        dist[0] = 0

        # Min-heap stores tuples of (distance, node)
        min_heap = [(0, 0)]
        while min_heap:
            d, curr = heapq.heappop(min_heap)

            if d > dist[curr]:
                continue

            for neighbour, distance in adj[curr]:
                if d + distance < dist[neighbour]:
                    dist[neighbour] = d + distance
                    heapq.heappush(min_heap, (d + distance, neighbour))

        # Phase 2: Count reachable nodes
        # 1. Reachable original nodes
        reachables = 0
        for i in range(n):
            if dist[i] <= maxMoves:
                reachables += 1

        # Reachable subdivided nodes on edges
        for u, v, cnt in edges:
            moves_from_u = max(0, maxMoves - dist[u])
            moves_from_v = max(0, maxMoves - dist[v])

            # Cap at 'cnt' so we don't double-count overlapping coverage
            reachables += min(cnt, moves_from_u + moves_from_v)

        return reachables
```

```python
solution = Solution()
```

```python
edges = [[0,1,10],[0,2,1],[1,2,2]]
maxMoves = 6
n = 3

# Expected: 13
print(solution.reachableNodes(edges, maxMoves, n))
```
