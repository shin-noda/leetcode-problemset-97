```python
class Solution:
    def findLength(self, nums1, nums2):
        n1 = len(nums1)
        n2 = len(nums2)
        max_len = 0

        # dp[i][j] stores the common prefix length starting at nums1[i] and nums2[j].
        # dp[i+1][j+1] means the number of common prefix elements in nums1[i+1:] and nums2[j+1:].
        dp = [[0] * (n2 + 1) for _ in range(n1 + 1)]

        # Traverse backward to build on already solved subproblems
        for i in range(n1 - 1, -1, -1):
            for j in range(n2 - 1, -1, -1):
                if nums1[i] == nums2[j]:
                    dp[i][j] = dp[i + 1][j + 1] + 1
                    max_len = max(max_len, dp[i][j])

        return max_len
```

```python
solution = Solution()
```

```python
nums1 = [1,2,3,2,1]
nums2 = [3,2,1,4,7]

print(solution.findLength(nums1, nums2))
```

```python
nums1 = [0,0,0,0,0]
nums2 = [0,0,0,0,0]

print(solution.findLength(nums1, nums2))
```
