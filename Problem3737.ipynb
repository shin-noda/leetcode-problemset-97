```python
class Solution:
    def countMajoritySubarrays(self, nums, target):
        # Use Sliding Window
        result = 0
        n = len(nums)

        for i in range(n):
            # Initialize the window value
            window = 0
            left = 0
            mid = (i + 1) // 2
            for w in range(i):
                if nums[w] == target:
                    window += 1

            for right in range(i, n):
                if nums[right] == target:
                    window += 1

                if window > mid:
                    result += 1

                if nums[left] == target:
                    window -= 1

                left += 1

        return result
```
