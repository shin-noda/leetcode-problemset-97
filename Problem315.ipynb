```python
class Solution:
    def merge(self, left, right, l_i, r_i):
        i = j = 0
        merged = []
        merged_indices = []

        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                merged.append(left[i])
                merged_indices.append(l_i[i])

                # Record the inversion for the origianl index
                self.counts[l_i[i]] += j
                i += 1

            else:
                merged.append(right[j])
                merged_indices.append(r_i[j])
                j += 1

        # We need to add 'j' to the rest of elements.
        while i < len(left):
            merged.append(left[i])
            merged_indices.append(l_i[i])
            self.counts[l_i[i]] += j
            i += 1

        merged.extend(right[j:])
        merged_indices.extend(r_i[j:])

        return merged, merged_indices


    def mergeSort(self, arr, indices):
        if len(arr) <= 1:
            return arr, indices

        mid = len(arr) // 2

        left, l_i = self.mergeSort(arr[:mid], indices[:mid])
        right, r_i = self.mergeSort(arr[mid:], indices[mid:])

        merged, merged_indices = self.merge(left, right, l_i, r_i)

        return merged, merged_indices


    def countSmaller(self, nums):
        # Idea: Use Merge Sort with Inversion Counting
        self.counts = [0] * len(nums)

        indices = []
        for i in range(len(nums)):
            indices.append(i)

        _, _ = self.mergeSort(nums, indices)

        return self.counts
```

```python
solution = Solution()
```

```python
nums = [5,2,6,1]

# Expected: [2,1,1,0]
print(solution.countSmaller(nums))
```

```python
nums = [-1]

# Expected: [0]
print(solution.countSmaller(nums))
```

```python
nums = [-1,-1]

# Expected: [0,0]
print(solution.countSmaller(nums))
```

```python
nums = [1,2,3,4,5]

# Expected: [0,0,0,0,0]
print(solution.countSmaller(nums))
```
