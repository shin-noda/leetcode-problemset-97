```python
class FenwickTree:
    def __init__(self, n):
        self.n = n
        self.tree = [0] * (n + 1)


    def update(self, index, delta):
        while index <= self.n:
            self.tree[index] += delta
            index += index & (-index)


    def query(self, index):
        total = 0
        while index > 0:
            total += self.tree[index]
            index -= index & (-index)

        return total


class Solution:
    def reversePairs(self, nums):
        # Step 1: Collect all values (nums[i] and 2 * nums[i])
        all_vals = set()
        for x in nums:
            all_vals.add(x)
            all_vals.add(2 * x)

        sorted_vals = sorted(all_vals)


        # Custom binary search helper to find 1-based rank
        def get_rank(target):
            left, right = 0, len(sorted_vals) - 1
            while left <= right:
                mid = (left + right) // 2
                if sorted_vals[mid] == target:
                    # Convert 0-index to 1-index rank
                    return mid + 1

                elif sorted_vals[mid] < target:
                    left = mid + 1

                else:
                    right = mid - 1

            return left + 1


        bit = FenwickTree(len(sorted_vals))
        count = 0

        # Step 2: Traverse right to left
        for x in reversed(nums):
            # Query count of elements in BIT strictly smaller than x (where 2 * y < x)
            rank_x = get_rank(x)
            count += bit.query(rank_x - 1)

            # Insert 2 * x into BIT
            rank_2x = get_rank(2 * x)
            bit.update(rank_2x, 1)

        return count
```

```python
solution = Solution()
```

```python
nums = [1,3,2,3,1]

print(solution.reversePairs(nums))
```
