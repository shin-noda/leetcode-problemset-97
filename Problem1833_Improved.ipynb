```python
class Solution:
    def maxIceCream(self, costs, coins):
        # Finding max cost dynamically or using the known constraint bound
        max_cost = max(costs)
        counts = [0] * (max_cost + 1)

        for cost in costs:
            counts[cost] += 1

        result = 0
        for cost in range(1, max_cost + 1):
            if counts[cost] == 0:
                continue

            # If we can't even afford 1 bar at this price, we're done
            if coins < cost:
                break

            # Take as many as we can afford or as many as exist
            take = min(counts[cost], coins // cost)
            result += take
            coins -= take * cost

        return result
```
