```python
class Node:
    def __init__(self):
        self.left = None
        self.right = None
        self.max_val = 0
        self.lazy = 0


class MyCalendar:
    def __init__(self):
        self.root = Node()


    def query(self, node, l, r, ql, qr):
        # Completely outside
        if qr <= l or r <= ql:
            return 0

        # Completely inside
        if ql <= l and r <= qr:
            return node.max_val

        mid = (l + r) // 2

        # Push lazy value to children
        self.push(node)

        left_max = self.query(node.left, l, mid, ql, qr)
        right_max = self.query(node.right, mid, r, ql, qr)

        return max(left_max, right_max)


    def update(self, node, l, r, ql, qr):
        # Completely outside
        if qr <= l or r <= ql:
            return

        # Completely inside
        if ql <= l and r <= qr:
            node.max_val += 1
            node.lazy += 1
            return

        mid = (l + r) // 2

        self.push(node)
        self.update(node.left, l, mid, ql, qr)
        self.update(node.right, mid, r, ql, qr)

        node.max_val = max(node.left.max_val, node.right.max_val)


    def push(self, node):
        if node.left is None:
            node.left = Node()

        if node.right is None:
            node.right = Node()

        if node.lazy:
            node.left.max_val += node.lazy
            node.left.lazy += node.lazy

            node.right.max_val += node.lazy
            node.right.lazy += node.lazy

            node.lazy = 0


    def book(self, startTime, endTime):
        # Check whether there is already a booking
        if self.query(self.root, 0, 10**9, startTime, endTime) > 0:
            return False

        # No overlap -> add the booking
        self.update(self.root, 0, 10**9, startTime, endTime)

        return True
```
