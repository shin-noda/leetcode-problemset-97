```python
class MyCalendar:
    def __init__(self):
        self.bookings = []


    def book(self, startTime, endTime):
        # Binary search for the insertion index (bisect_right logic)
        left, right = 0, len(self.bookings)
        while left < right:
            mid = (left + right) // 2
            if self.bookings[mid][0] <= startTime:
                left = mid + 1

            else:
                right = mid

        idx = left

        # Check overlap with previous event
        if idx > 0 and self.bookings[idx - 1][1] > startTime:
            return False

        # Check overlap with next event
        if idx < len(self.bookings) and self.bookings[idx][0] < endTime:
            return False

        self.bookings.insert(idx, (startTime, endTime))

        return True
```
