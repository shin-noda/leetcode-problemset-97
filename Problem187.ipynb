```python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.count = 0


class Solution:
    def findRepeatedDnaSequences(self, s):
        sequences = []
        n = len(s)
        n_dna = 10
        root = TrieNode()

        for i in range(n - n_dna + 1):
            # For each inner loop, you start from the root
            curr = root

            for j in range(n_dna):
                if s[i + j] not in curr.children:
                    curr.children[s[i + j]] = TrieNode()

                # Move on to the child node
                curr = curr.children[s[i + j]]

            # Check if the count is exactly 1. If it's greater than 1,
            # it's already added to the sequence.
            if curr.count == 1:
                sequences.append(s[i:i + n_dna])

            # Increment the count here, not before the checking.
            curr.count += 1

        return sequences
```

```python
solution = Solution()
```

```python
s = "AAAAACCCCCAAAAACCCCCCAAAAAGGGTTT"

print(solution.findRepeatedDnaSequences(s))
```
