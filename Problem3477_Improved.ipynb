```python
class SegmentTree:
    def __init__(self, baskets):
        self.n = len(baskets)
        self.tree = [0] * (4 * self.n)
        self.build(1, 0, self.n - 1, baskets)


    def build(self, node, l, r, baskets):
        if l == r:
            self.tree[node] = baskets[l]
            return

        mid = (l + r) // 2

        self.build(2 * node, l, mid, baskets)
        self.build(2 * node + 1, mid + 1, r, baskets)

        self.tree[node] = max(self.tree[2 * node], self.tree[2 * node + 1])


    def find_leftmost(self, fruit):
        # No basket can hold this fruit
        if self.tree[1] < fruit:
            return -1

        return self._find_leftmost(1, 0, self.n - 1, fruit)


    def _find_leftmost(self, node, l, r, fruit):
        if l == r:
            return l

        mid = (l + r) // 2

        # If the left segment contains a basket that can hold the fruit, go left.
        if self.tree[2 * node] >= fruit:
            return self._find_leftmost(2 * node, l, mid, fruit)

        # Otherwise, go right.
        return self._find_leftmost(2 * node + 1, mid + 1, r, fruit)


    def update(self, index):
        self._update(1, 0, self.n - 1, index)


    def _update(self, node, l, r, index):
        if l == r:
            self.tree[node] = 0
            return

        mid = (l + r) // 2

        if index <= mid:
            self._update(2 * node, l, mid, index)

        else:
            self._update(2 * node + 1, mid + 1, r, index)

        self.tree[node] = max(self.tree[2 * node], self.tree[2 * node + 1])


class Solution:
    def numOfUnplacedFruits(self, fruits, baskets):
        segment_tree = SegmentTree(baskets)
        answer = 0

        for fruit in fruits:
            index = segment_tree.find_leftmost(fruit)

            if index == -1:
                answer += 1

            else:
                segment_tree.update(index)

        return answer
```

```python
solution = Solution()
```

```python
fruits = [4,2,5]
baskets = [3,5,4]

print(solution.numOfUnplacedFruits(fruits, baskets))
```
