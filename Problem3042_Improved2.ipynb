```python
class TrieNode:
    def __init__(self):
        # Maps (prefix_char, suffix_char) -> TrieNode
        self.children = {}

        # How many words end exactly at this node
        self.count = 0


class Solution:
    def countPrefixSuffixPairs(self, words):
        # Use Trie
        # Time Complexity: O(n * l) where n = len(words) and l in max(words)
        root = TrieNode()
        total_pairs = 0

        for w in words:
            n = len(w)
            curr = root

            # Walk/Insert 'w' into the Trie by character pairs
            for k in range(n):
                pair = (w[k], w[n - 1 - k])

                if pair not in curr.children:
                    curr.children[pair] = TrieNode()

                curr = curr.children[pair]

                # If a previous shorter word ended at this exact node,
                # it is both a prefix and a suffix of 'w'
                total_pairs += curr.count

            # Mark that 'w' ends here
            curr.count += 1

        return total_pairs
```

```python
solution = Solution()
```

```python
words = ["a","aba","ababa","aa"]

# Expected: 4
print(solution.countPrefixSuffixPairs(words))
```

```python
words = ["pa","papa","ma","mama"]

# Expected: 2
print(solution.countPrefixSuffixPairs(words))
```
