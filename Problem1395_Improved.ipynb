```python
class FenwickTree:
    def __init__(self, size):
        self.tree = [0] * (size + 1)


    def update(self, index, delta):
        while index < len(self.tree):
            self.tree[index] += delta
            index += index & (-index)


    def query(self, index):
        total = 0
        while index > 0:
            total += self.tree[index]
            index -= index & (-index)

        return total


class Solution:
    def numTeams(self, rating):
        n = len(rating)

        # Coordinate compression
        sorted_ranks = {val: i + 1 for i, val in enumerate(sorted(rating))}
        ranks = [sorted_ranks[r] for r in rating]

        left_bit = FenwickTree(n)
        right_bit = FenwickTree(n)

        # Populate right_bit with all elements initially
        for r in ranks:
            right_bit.update(r, 1)

        teams = 0

        for j, r in enumerate(ranks):
            # Remove current element from right BIT before treating it as middle
            right_bit.update(r, -1)

            left_less = left_bit.query(r - 1)
            left_greater = j - left_less

            right_less = right_bit.query(r - 1)
            right_greater = (n - 1 - j) - right_less

            teams += (left_less * right_greater) + (left_greater * right_less)

            # Add current element to left BIT for future iterations
            left_bit.update(r, 1)

        return teams
```

```python
solution = Solution()
```

```python
rating = [2,5,3,4,1]

print(solution.numTeams(rating))
```
