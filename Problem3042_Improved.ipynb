```python
class Solution:
    def countPrefixSuffixPairs(self, words):
        # Use Rolling Hash + Verification
        # Time Complexity: O(n * l) where n = len(words) and l in max(words)
        BASE = 31
        MOD = 10**9 + 7

        # (length, hash) -> list of complete previous words
        seen = {}
        answer = 0

        for w in words:
            n = len(w)
            prefix_hash = 0
            suffix_hash = 0
            power = 1

            # Enumerate every prefix/suffix pair
            for k in range(n):
                # Prefix hash
                p = ord(w[k]) - ord('a') + 1
                prefix_hash = (prefix_hash * BASE + p) % MOD

                # Suffix hash
                s = ord(w[n - 1 - k]) - ord('a') + 1
                suffix_hash = (suffix_hash + s * power) % MOD
                power = (power * BASE) % MOD

                # Possible border
                if prefix_hash == suffix_hash:
                    border = w[:k + 1]

                    # Verify to eliminate hash collisions
                    if border == w[n - 1 - k:]:
                        key = (k + 1, prefix_hash)

                        if key in seen:
                            for prev in seen[key]:
                                if prev == border:
                                    answer += 1

            # Insert complete word after querying
            full_hash = 0
            for ch in w:
                full_hash = (full_hash * BASE + (ord(ch) - ord('a') + 1)) % MOD

            key = (n, full_hash)
            if key not in seen:
                seen[key] = []

            seen[key].append(w)

        return answer
```

```python
solution = Solution()
```

```python
words = ["a","aba","ababa","aa"]

# Expected: 4
print(solution.countPrefixSuffixPairs(words))
```
