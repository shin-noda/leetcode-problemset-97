```python
from collections import deque

class Solution:
    def possibleBipartition(self, n, dislikes):
        # Note: elements in 'dislikes" are 1-indexed, so we have to shift by 1
        colour = [-1] * n
        dq = deque()

        # Create a graph
        adj = [[] for _ in range(n)]
        for a, b in dislikes:
            # We convert 1-index to 0-index
            adj[a - 1].append(b - 1)
            adj[b - 1].append(a - 1)

        for start in range(n):
            # Check if the node is visited
            if colour[start] != -1:
                continue

            colour[start] = 0
            dq.append(start)

            while dq:
                node = dq.popleft()

                for neighbour in adj[node]:
                    if colour[neighbour] == -1:
                        colour[neighbour] = 1 - colour[node]
                        dq.append(neighbour)

                    elif colour[neighbour] == colour[node]:
                        return False

        return True
```

```python
solution = Solution()
```

```python
n = 4
dislikes = [[1,2],[1,3],[2,4]]

print(solution.possibleBipartition(n, dislikes))
```
