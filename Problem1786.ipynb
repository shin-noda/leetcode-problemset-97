```python
import heapq
from collections import defaultdict

class Solution:
    def countRestrictedPaths(self, n, edges):
        # This is a DAG (Directed Acyclic Graph) problem.
        MOD = 10**9 + 7

        # Build adjacency list
        adj = defaultdict(list)
        for u, v, w in edges:
            adj[u].append((v, w))
            adj[v].append((u, w))

        # Single-Source Dijkstra starting from node n
        dist = [float('inf')] * (n + 1)
        dist[n] = 0

        # (distance, current_node)
        min_heap = [(0, n)]

        while min_heap:
            d, curr = heapq.heappop(min_heap)

            if d > dist[curr]:
                continue

            for neighbour, weight in adj[curr]:
                if dist[curr] + weight < dist[neighbour]:
                    dist[neighbour] = dist[curr] + weight
                    heapq.heappush(min_heap, (dist[neighbour], neighbour))

        # DP with Memoization (DFS) from node 1 to node n
        memo = {}

        def dfs(curr):
            # Base Case: reached target not n
            if curr == n:
                return 1

            if curr in memo:
                return memo[curr]

            total_paths = 0
            for neighbour, _ in adj[curr]:
                # Only traverse if it satisfies the restricted condition
                if dist[curr] > dist[neighbour]:
                    total_paths = (total_paths + dfs(neighbour)) % MOD

            memo[curr] = total_paths

            return memo[curr]

        return dfs(1)
```

```python
solution = Solution()
```

```python
n = 5
edges = [[1,2,3],[1,3,3],[2,3,1],[1,4,2],[5,2,2],[3,5,1],[5,4,10]]

print(solution.countRestrictedPaths(n, edges))
```
