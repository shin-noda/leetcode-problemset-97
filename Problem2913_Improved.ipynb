```python
class SegmentTree:
    def __init__(self, n):
        self.n = n

        # Stores sum of d
        self.sum = [0] * (4 * n)

        # Stores sum of d^2
        self.sq_sum = [0] * (4 * n)

        # Lazy propagation tag
        self.lazy = [0] * (4 * n)


    def _push(self, node, l, r):
        if self.lazy[node] != 0:
            val = self.lazy[node]
            mid = (l + r) // 2

            # Update left child
            self._apply(2 * node, l, mid, val)

            # Update right child
            self._apply(2 * node + 1, mid + 1, r, val)

            self.lazy[node] = 0


    def _apply(self, node, l, r, val):
        length = r - l + 1

        # (d + val)^2 = d^2 + 2 * val * d + val^2
        self.sq_sum[node] += 2 * val * self.sum[node] + (val ** 2) * length
        self.sum[node] += val * length
        self.lazy[node] += val


    def update(self, node, l, r, ql, qr, val):
        if ql <= l and r <= qr:
            self._apply(node, l, r, val)
            return

        self._push(node, l, r)
        mid = (l + r) // 2

        if ql <= mid:
            self.update(2 * node, l, mid, ql, qr, val)

        if qr > mid:
            self.update(2 * node + 1, mid + 1, r, ql, qr, val)

        self.sum[node] = self.sum[2 * node] + self.sum[2 * node + 1]
        self.sq_sum[node] = self.sq_sum[2 * node] + self.sq_sum[2 * node + 1]


class Solution:
    def sumCounts(self, nums):
        # Idea: Segment Tree with Lazy Propagation
        n = len(nums)
        tree = SegmentTree(n)
        last_pos = {}
        total_sum = 0

        for j, val in enumerate(nums):
            prev = last_pos.get(val, -1)

            # Increment d(i, j) for all i in [prev + 1, j]
            tree.update(1, 0, n - 1, prev + 1, j, 1)

            # Accumulate sum of squares for all subarrays ending at j
            total_sum += tree.sq_sum[1]
            last_pos[val] = j

        return total_sum
```

```python
solution = Solution()
```

```python
nums = [1,2,1]

print(solution.sumCounts(nums))
```
