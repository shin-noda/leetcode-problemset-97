```python
class Solution:
    def eventualSafeNodes(self, graph):
        # A terminal node: No outgoing edges
        # A safe node: Every possible path starting from that node leads to a terminal node

        terminal_nodes = set()
        safe_nodes = set()
        dir = {}
        rev = {}

        for i in range(len(graph)):
            if len(graph[i]) == 0:
                terminal_nodes.add(i)
                safe_nodes.add(i)

            else:
                dir[i] = set()
                for j in range(len(graph[i])):
                    dir[i].add(graph[i][j])

                    if graph[i][j] not in rev:
                        rev[graph[i][j]] = set()

                    rev[graph[i][j]].add(i)

        while terminal_nodes:
            node = terminal_nodes.pop()

            if node not in rev:
                continue

            nodes = rev[node]

            # Delete the key
            rev.pop(node, None)

            for curr in nodes:
                if curr in dir:
                    dir[curr].discard(node)

                    if len(dir[curr]) == 0:
                        safe_nodes.add(curr)
                        terminal_nodes.add(curr)

                        # Delete the key
                        dir.pop(curr, None)

        return [i for i in range(len(graph)) if i in safe_nodes]
```

```python
solution = Solution()
```

```python
graph = [[1,2],[2,3],[5],[0],[5],[],[]]

print(solution.eventualSafeNodes(graph))
```
