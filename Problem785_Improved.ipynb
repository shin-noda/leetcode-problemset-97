```python
from collections import deque

class Solution:
    def isBipartite(self, graph):
        # Bipartite: the nodes can be partitioned into two independent sets A and B
        # such that every edge in the graph connects a node in set A and a node in set B.

        # Idea: Separate nodes into two groups and for each group, nodes within the
        # group shouldn't have any connections

        n = len(graph)
        colour = [-1] * n

        for start in range(n):
            if colour[start] != -1:
                continue

            colour[start] = 0
            queue = deque([start])

            while queue:
                node = queue.popleft()

                for neighbour in graph[node]:
                    # If color[neighbor] is not -1 and not the same as color[node], it'll be ignored.
                    if colour[neighbour] == -1:
                        # colour[neighbour] is either 0 or 1.
                        colour[neighbour] = 1 - colour[node]
                        queue.append(neighbour)

                    elif colour[neighbour] == colour[node]:
                        return False

        return True
```

```python
solution = Solution()
```

```python
graph = [[1,2,3],[0,2],[0,1,3],[0,2]]

print(solution.isBipartite(graph))
```

```python
graph = [[1,3],[0,2],[1,3],[0,2]]

print(solution.isBipartite(graph))
```

```python
graph = [[1],[0,3],[3],[1,2]]

# Expected: True
print(solution.isBipartite(graph))
```

```python
graph = [[3],[2,4],[1],[0,4],[1,3]]

# Expected: True
print(solution.isBipartite(graph))
```
