```python
class Solution:
    def arrayPairSum(self, nums):
        # Use Counting Sort
        # Time Complexity: O(n + K) where n = len(nums) and K = 20001
        OFFSET = 10**4
        k = 20001

        # Frequency array for values in range [-10**4, 10**4]
        count = [0] * k
        for num in nums:
            count[num + OFFSET] += 1

        total_sum = 0

        # Tracks whether we are at an even index in sorted order
        take = True

        for val in range(k):
            while count[val] > 0:
                if take:
                    # We do - OFFSET because all values are shifted by OFFSET.
                    total_sum += val - OFFSET

                # Toggle state: the next element is paired/skipped
                take = not take

                # Reduce the count in count[val]
                count[val] -= 1

        return total_sum
```
