```python
class Solution:
    def countRectangles(self, rectangles, points):
        # Array of 101 empty lists to store lengths for each height h (0 to 100)
        height_to_lengths = [[] for _ in range(101)]

        for l, h in rectangles:
            height_to_lengths[h].append(l)

        # Sort each non-empty length list
        for h in range(101):
            if height_to_lengths[h]:
                height_to_lengths[h].sort()

        result = []

        for x, y in points:
            count = 0

            # Check all possible valid heights from y up to 100
            for h in range(y, 101):
                lengths = height_to_lengths[h]

                if not lengths:
                    continue

                # Binary Search
                left, right = 0, len(lengths)

                while left < right:
                    mid = (left + right) // 2

                    if lengths[mid] < x:
                        left = mid + 1

                    else:
                        right = mid

                # Number of rectangles with height 'h' and length >= x
                count += len(lengths) - left

            result.append(count)

        return result
```

```python
solution = Solution()
```

```python
rectangles = [[1,2],[2,3],[2,5]]
points = [[2,1],[1,4]]

print(solution.countRectangles(rectangles, points))
```
