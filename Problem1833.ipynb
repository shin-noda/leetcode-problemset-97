```python
class Solution:
    def maxIceCream(self, costs, coins):
        RANGE = 100001
        n = len(costs)
        index = 0
        checked = 0
        result = 0

        counts = [0] * RANGE
        for cost in costs:
            counts[cost] += 1

        while coins > 0 and checked < n:
            while index < RANGE and counts[index] == 0:
                index += 1

            # Short circuit
            if coins < index:
                break

            # Pay for an ice cream bar
            coins -= index

            # Reduce the stock of ice cream bars by 1
            counts[index] -= 1

            checked += 1
            result += 1

        return result
```
