```python
class Solution:
    def numTeams(self, rating):
        # Idea: DP
        # Let's say rating[i] < rating[j] and
        # there are say 5 combinations that rating[j] < rating[k],
        # then there are 5 combinations that meet
        # rating[i] < rating[j] < rating[k].
        answer = 0
        n = len(rating)

        # count1 and count2 track for increasing and decreasing sequences respectively
        count1 = [0] * n
        count2 = [0] * n

        # Compute for the count table
        for i in range(1, n):
            for j in range(i + 1, n):
                if rating[i] < rating[j]:
                    count1[i] += 1

                elif rating[i] > rating[j]:
                    count2[i] += 1

        for i in range(n):
            for j in range(i + 1, n):
                if rating[i] < rating[j]:
                    answer += count1[j]

                elif rating[i] > rating[j]:
                    answer += count2[j]

        return answer
```

```python
solution = Solution()
```

```python
rating = [2,5,3,4,1]

print(solution.numTeams(rating))
```
