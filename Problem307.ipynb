```python
class NumArray:
    def __init__(self, nums):
        self.n = len(nums)
        self.nums = nums
        self.sum = [0] * (self.n + 1)

        for i in range(self.n):
            self._initialize(i, self.nums[i])


    def _initialize(self, index, val):
        index += 1
        while index <= self.n:
            self.sum[index] += val
            index += index & (-index)


    def update(self, index, val):
        # Calculate the difference of the current num[index] - val
        # Note: here, index is 0-index
        diff = val - self.nums[index]

        # Update the value of nums[index] (still 0-index)
        self.nums[index] = val

        # Convert 0-index to 1-index
        index += 1

        while index <= self.n:
            self.sum[index] += diff
            index += index & (-index)


    def sumRange(self, left, right):
        # Note: we want the sum value including 'left', so we subtract -1
        return self._query(right) - self._query(left - 1)


    def _query(self, index):
        # Convert 0-index to 1-index
        index += 1
        total = 0
        while index > 0:
            total += self.sum[index]
            index -= index & (-index)

        return total
```
