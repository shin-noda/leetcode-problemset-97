```python
class Solution:
    def findNumberOfLIS(self, nums):
        # Idea: Use DP to solve the problem with O(N^2) time complexity
        n = len(nums)

        if n == 0:
            return 0

        length = [1] * n
        count = [1] * n
        max_len = 1

        for i in range(n):
            for j in range(i):
                if nums[j] < nums[i]:
                    if length[j] + 1 > length[i]:
                        length[i] = length[j] + 1
                        count[i] = count[j]

                    elif length[j] + 1 == length[i]:
                        count[i] += count[j]

            max_len = max(max_len, length[i])

        # Sum the counts of all indices that achieve the global maximum LIS length
        return sum(c for l, c in zip(length, count) if l == max_len)
```

```python
solution = Solution()
```

```python
nums = [1,3,5,4,7]

print(solution.findNumberOfLIS(nums))
```

```python
nums = [2,2,2,2,2]

print(solution.findNumberOfLIS(nums))
```
