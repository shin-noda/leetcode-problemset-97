```python
from collections import deque

class Solution:
    def isBipartite(self, graph):
        # Bipartite: the nodes can be partitioned into two independent sets A and B
        # such that every edge in the graph connects a node in set A and a node in set B.

        # Idea: Separate nodes into two groups and for each group, nodes within the
        # group shouldn't have any connections

        group0 = set()
        group1 = set()
        n = len(graph)

        adj = {}
        dq = deque([])
        for i in range(n):
            # We ignore isolated nodes
            if graph[i]:
                adj[i] = graph[i]

                if len(dq) == 0:
                    dq.append(i)

        group = 0
        while adj:
            num_el = len(dq)

            # We need to pick an arbiturary node.
            if num_el == 0:
                dq.append(next(iter(adj)))
                num_el = 1

            for i in range(num_el):
                el = dq.popleft()

                if el not in adj:
                    continue

                items = adj.pop(el)

                for item in items:
                    if group == 0:
                        if item in group0:
                            return False

                        if item in group1:
                            continue

                        group1.add(item)

                    else:
                        if item in group1:
                            return False

                        if item in group0:
                            continue

                        group0.add(item)

                    dq.append(item)

            # Switch a group
            group = 1 ^ group

        return True
```

```python
solution = Solution()
```

```python
graph = [[1,2,3],[0,2],[0,1,3],[0,2]]

print(solution.isBipartite(graph))
```

```python
graph = [[1,3],[0,2],[1,3],[0,2]]

print(solution.isBipartite(graph))
```
