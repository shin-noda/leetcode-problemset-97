```python
class Solution:
    def numTrees(self, n):
        # A Binary Search Tree satisfies the following properties:
        # -All nodes in the left subtree of a node contain values strictly less than the node’s value.
        # -All nodes in the right subtree of a node contain values strictly greater than the node’s value.

        # Idea:
        # From dp[i] to dp[i + 1], you can add a new node in at most 2 ways
        # for each BST.
        #
        # Nope :)
        #
        # For each configuration, there are 4 ways to add one new node:
        # top-left, top-right, bottom-left, and bottom-right.
        # Not all configurations are valid.
        #
        # Should I think in terms of symmetry?
        # It seems like I can find the mirror image around the vertical line.
        #
        # Use Backtracking instead.
        #
        # Nope :)
        #
        # What's the subproblme in this question?
        #
        # The subproblem is for each left and right part of BST, how many
        # sub BSTs are there.

        # Base cases
        if n <= 1:
            return 1

        dp = [0] * (n + 1)

        # There are 1 way to make BST with 0 node and 1 node
        dp[0] = 1
        dp[1] = 1

        for i in range(2, n + 1):
            for left in range(i):
                # Since there is 1 root node, left + right = i - 1.
                # This '-1' is for subtracting the root.
                right = i - left - 1
                dp[i] += dp[left] * dp[right]

        return dp[n]
```

```python
solution = Solution()
```
